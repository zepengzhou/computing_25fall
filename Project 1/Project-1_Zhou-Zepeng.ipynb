Try clicking the title to get into the detailed page of each drama

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import pandas as pd

# --- 1) 新建 driver（全新会话）---
opts = Options()
# 可选：无头模式更稳
# opts.add_argument("--headless=new")
opts.add_argument("--no-sandbox")
opts.add_argument("--disable-dev-shm-usage")

# 如果你使用 webdriver-manager，可以这样（但需能联网下载）：
# from webdriver_manager.chrome import ChromeDriverManager
# service = Service(ChromeDriverManager().install())
# driver = webdriver.Chrome(service=service, options=opts)

# 如果你本地已配好 chromedriver（在 PATH 中），直接：
driver = webdriver.Chrome(options=opts)
wait = WebDriverWait(driver, 15)

try:
    # --- 2) 打开列表页 ---
    drama_url = "https://www.imdb.com/list/ls548976736/"
    driver.get(drama_url)

    wait.until(EC.presence_of_all_elements_located(
        (By.CSS_SELECTOR, "a.ipc-title-link-wrapper")
    ))
    title_link_els = driver.find_elements(By.CSS_SELECTOR, "a.ipc-title-link-wrapper")
    detail_urls = [el.get_attribute("href") for el in title_link_els[:10]]

    results = []

    # --- 3) 逐个进入详情页抓发布日期 ---
    for url in detail_urls:
        driver.get(url)

        # 标题
        try:
            title = wait.until(EC.presence_of_element_located(
                (By.CSS_SELECTOR, "h1[data-testid='hero__pageTitle']"))
            ).text
        except:
            title = driver.find_element(By.TAG_NAME, "h1").text

        # release date（只选右侧日期的 a）
        try:
            li = wait.until(EC.presence_of_element_located(
                (By.CSS_SELECTOR, "li[data-testid='title-details-releasedate']"))
            )
            driver.execute_script("arguments[0].scrollIntoView({block:'center'});", li)
            time.sleep(0.3)

            a_node = wait.until(EC.visibility_of_element_located((
                By.CSS_SELECTOR,
                "li[data-testid='title-details-releasedate'] a.ipc-metadata-list-item__list-content-item"
            )))
            release_date = a_node.get_attribute("textContent").strip()
        except Exception as e:
            release_date = "N/A"

        results.append({"title": title, "release_date": release_date, "url": url})
        print(f"[OK] {title} -> {release_date}")


df = pd.DataFrame(results)
df
```
